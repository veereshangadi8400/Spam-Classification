```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_email_contents, y_labels, test_size=0.2, random_state=42)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
countVectorizer = CountVectorizer()
countVectorizer
```

```python
X_train_vectorized = countVectorizer.fit_transform(X_train)
X_train_vectorized
```

```python
countVectorizer.vocabulary_
```

```python
X_test_vectorized = countVectorizer.transform(X_test)
X_test_vectorized
```

```python
from sklearn.naive_bayes import MultinomialNB
multinomialNB = MultinomialNB()
multinomialNB.fit(X_train_vectorized, y_train)
```

```python
NB_predictions = multinomialNB.predict(X_test_vectorized)
NB_predictions
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score
NB_precision = precision_score(y_test, NB_predictions)
NB_precision
```

```python
NB_recall = recall_score(y_test, NB_predictions)
NB_recall
```

```python
NB_f1score = f1_score(y_test, NB_predictions)
NB_f1score
```

```python
from sklearn.metrics import confusion_matrix
NB_confusion = confusion_matrix(y_test, NB_predictions)
NB_confusion
```
